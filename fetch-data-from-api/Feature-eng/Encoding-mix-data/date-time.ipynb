Working with data

```python
date['date'] = pd.to_datetime(date['date']) # so that apde date time functions apply kari sakiye so changing the data type of date col of date data
```

```python
date.info()
```

```python
date['date_year'] = date['date'].dt.year
```

```python
date.sample(5)
```

```python
date['date-month'] = date['date'].dt.month
```

```python
date['date-month-name'] = date['date'].dt.month_name()
```

```python
date['date-day'] = date['date'].dt.day
```

```python
date['date-dow'] = date['date'].dt.dayofweek
```

```python
date['date-dayname'] = date['date'].dt.day_name()
```

```python
date['is_weekend'] = np.where(date['date-dayname'].isin(['Sunday','Saturday']),1,0)
```

```python
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['date_week'] = date['date'].dt.isocalendar().week
```

```python
date['quarter'] =  date['date'].dt.quarter
```

```python
date['semester'] = np.where(date['quarter'].isin([1,2]),1,2)
```

```python
import datetime
```

```python
today = datetime.datetime.today()
```

```python
today - date['date']
```

```python
(today-date['date']).dt.days
```

```python
# Months passed
# (Year difference * 12) + Month difference

np.round((today.year - date['date'].dt.year) * 12 + (today.month - date['date'].dt.month), 0)
```
