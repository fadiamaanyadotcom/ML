## Making own LR Class

```python
class MeraLR:
    def __init__(self):
        self.coef_ = None
        self.intercept_ = None
    def fit(self,X_train,y_train):
        X_train = np.insert(X_train,0,1,axis = 1) # insert kaya dataset ma then comes kayu index then what value and axis = 1 aetle col so 1st col of matrix ma value = 1 mukvanu che
        betas = np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(y_train)
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]
    def predict(self,X_test):
        y_pred = np.dot(X_test,self.coef_) + self.intercept_
        return y_pred
```

```python
lr = MeraLR()
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```
