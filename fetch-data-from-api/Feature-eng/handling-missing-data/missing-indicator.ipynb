```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer,MissingIndicator
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
si = SimpleImputer()
X_train_trf = si.fit_transform(X_train)# so jya null hase tya by default mean thase and model ae rete train
X_test_trf = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()
clf.fit(X_train_trf,y_train)
y_pred = clf.predict(X_test_trf)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

## Missing Indicator

so ae su hoy k aek col ma lets say koi NA value che to aek seperate col banse
like
age ha 3 values and 1 NA
age_missing will be made which will have 3 False and 1 true

so models know how to diff between null and not null values

this doesnt work always
but if accuracy or r2 score is less then this is must try

```python
mi = MissingIndicator()
si = SimpleImputer()

X_train_age_na = mi.fit_transform(X_train)
X_test_age_na = mi.transform(X_test)

X_train['Age_NA'] = X_train_age_na
X_test['Age_NA'] = X_test_age_na

X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)

clf.fit(X_train_trf2, y_train)
y_pred = clf.predict(X_test_trf2)

from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

## OR

```python
si = SimpleImputer(add_indicator=True)
X_train_trf3 = si.fit_transform(X_train)
X_test_trf3 = si.transform(X_test)

from sklearn.linear_model import LogisticRegression
clf = LogisticRegression()
clf.fit(X_train_trf3,y_train)
y_pred = clf.predict(X_test_trf3)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
