```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Pclass','Fare','Survived'])
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
# aapde weights neighbours ne change karine accuracy jovani 
knn = KNNImputer(n_neighbors=3,weights='distance')

X_train_trf = knn.fit_transform(X_train)
X_test_trf = knn.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train_trf,y_train)
y_pred = lr.predict(X_test_trf)

accuracy_score(y_test,y_pred)
```

```python
# Comparision
lr = LogisticRegression()
si = SimpleImputer()
X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)

lr.fit(X_train_trf2,y_train)
y_pred = lr.predict(X_test_trf2)

accuracy_score(y_test,y_pred)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_score

pipe_knn = Pipeline([
    ('imputer',KNNImputer(n_neighbors=3,weights='distance')),
    ('model',LogisticRegression())
])
pipe_simple = Pipeline([
    ('imputer',SimpleImputer()),
    ('model',LogisticRegression())
])

knn_scores = cross_val_score(pipe_knn,X,y,cv=5)
simple_scores = cross_val_score(pipe_simple,X,y,cv=5)

print(knn_scores.mean(),simple_scores.mean())
```
