MICE = multivariate imputation by chained equations

there are 3 types of missing data assumptions
1. MCAR ==> missing completely at random
            bija rows thi aeni val fill na thay and data is still raw...
2. MAR ==> missing at random
           bija rows thi aeni values fill thay i.e data has some corelation
3. MNAR ==> missing not at random
            data thi hathe karine values remove thyi che

We can use MICE in all 3 cases but best result in MAR

ADVANTAGE 
1. Accurate

DISADVANTAGE
1. slow

When u have multiple numerical values to handle like price temp etc rather than discrete categorical then its a regression problem

Regression: Predicts a number ("How much?").

Classification: Predicts a category ("Which class?").

Linear RegressionPurpose: Predicts a continuous quantity.
Output: Any real number (from negative infinity to positive infinity).

Logistic RegressionPurpose: Predicts the probability of a categorical outcome.
Output: A value between 0 and 1 (which maps to a class like "Yes" or "No").
