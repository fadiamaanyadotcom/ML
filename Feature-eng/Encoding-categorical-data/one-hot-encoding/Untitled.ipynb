## Dummy Variable Trap

there are 3 categories/colours yellow,blue,red and we cant rank it like 0, 1, 2 in categories like we did in ordinal-label-encoding or else the model will think one colour is more imp than other

so we represent in binary so if colour iis yellow then it is 1 and rest 2 cols or colours are 0

this is known as Dummy Variable and it leads to dummy variable trap as then the input cols become dependent to each other and in ml we have to make all the input cols independent of eachother

so for n cols we drop the 1st col as if 2 colours are 0 then ofc the 3rd colour is the correct colour

when there are many many cols/categories we take the categories that are frequently used and put others in diff category like others and then aena Dummy Variable banse
