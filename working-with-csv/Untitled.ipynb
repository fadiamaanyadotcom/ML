## Index_col parameter

## use cols_parameter
when uk what cols u dont want so while importing data you can import only specific cols


```python
pd.read_csv('aug_test.csv',usecols=['city','gender'])
```

## Squeeze Parameter
when we need 1 col(rarely) we use squeeze() so that we get pandas series not dataframe

```python
pd.read_csv('aug_test.csv', usecols=['city']).squeeze()
```

## Skip rows/n rows Parameter

```python
pd.read_csv('aug_test.csv',skiprows=[1,3,5,7]) 
# 1 is the 1st row index 0
```

```python
pd.read_csv('aug_test.csv')
```

```python
pd.read_csv('aug_test.csv', skiprows=lambda x: x % 2 != 0)
```

```python
pd.read_csv('aug_test.csv',nrows=100)
# nrows give the no. of rows we type here 100 so we get 100 rows
```
